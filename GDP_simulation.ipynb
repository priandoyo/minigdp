```python
# Simple Econometric Model: Predict GDP Growth from Investment Growth

import numpy as np
import pandas as pd
import statsmodels.api as sm

# Generate synthetic data for 10 years
np.random.seed(42)
years = np.arange(2010, 2020)
investment_growth = np.random.normal(0.05, 0.02, size=10)  # around 5% avg investment growth

# Assume true relationship: GDP_growth = 0.01 + 0.8 * Investment_growth + noise
true_alpha = 0.01
true_beta = 0.8
noise = np.random.normal(0, 0.01, size=10)
gdp_growth = true_alpha + true_beta * investment_growth + noise

# Create DataFrame
data = pd.DataFrame({
    'Year': years,
    'Investment_Growth': investment_growth,
    'GDP_Growth': gdp_growth
})

print("Sample data:")
print(data)

# Prepare regression inputs
X = sm.add_constant(data['Investment_Growth'])  # Adds intercept term
y = data['GDP_Growth']

# Fit OLS regression
model = sm.OLS(y, X).fit()

print("\nRegression Summary:")
print(model.summary())

# Predict GDP growth if investment growth is 6%
new_investment = 0.06
# Create a DataFrame with the same columns as X (const and Investment_Growth)
X_new = pd.DataFrame({'const': 1, 'Investment_Growth': new_investment}, index=[0])

predicted_gdp = model.predict(X_new)[0]
print(f"\nPredicted GDP growth with investment growth of 6%: {predicted_gdp:.4f}")

```
